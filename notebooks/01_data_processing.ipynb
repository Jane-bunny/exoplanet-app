# 01 - Data Processing

In this notebook, we will begin the processing pipeline for the exoplanet project.  
The goals are:
- Inspect the raw data files.
- Load the data into pandas DataFrames.
- Apply initial cleaning (remove nulls, format columns, handle units).
- Save a processed version into `data/processed/`.

This will make the dataset ready for visualization and modeling.

```python
print("\nKepler info:")
df_kepler.info()

print("\nTESS info:")
df_tess.info()
```

```python
# Clean Kepler dataset inline

kepler_cols = {
    "kepid": "star_id",
    "kepoi_name": "planet_id",
    "koi_disposition": "disposition",
    "koi_period": "orbital_period",
    "koi_duration": "transit_duration",
    "koi_depth": "transit_depth",
    "koi_prad": "planet_radius",
    "koi_teq": "equilibrium_temp",
    "koi_insol": "insolation_flux",
    "koi_model_snr": "snr",
    "koi_steff": "stellar_teff",
    "koi_slogg": "stellar_logg",
    "koi_srad": "stellar_radius",
    "ra": "ra",
    "dec": "dec",
    "koi_kepmag": "stellar_mag"
}

df_kepler_clean = df_kepler[list(kepler_cols.keys())].rename(columns=kepler_cols)
df_kepler_clean.head()
```

```python
# Clean TESS dataset inline

tess_cols = {
    "toi": "planet_id",
    "tid": "star_id",
    "tfopwg_disp": "disposition",
    "pl_orbper": "orbital_period",
    "pl_trandurh": "transit_duration",
    "pl_trandep": "transit_depth",
    "pl_rade": "planet_radius",
    "pl_eqt": "equilibrium_temp",
    "pl_insol": "insolation_flux",
    "st_teff": "stellar_teff",
    "st_logg": "stellar_logg",
    "st_rad": "stellar_radius",
    "ra": "ra",
    "dec": "dec",
    "st_tmag": "stellar_mag",
    "st_dist": "stellar_dist"
}

df_tess_clean = df_tess[list(tess_cols.keys())].rename(columns=tess_cols)
df_tess_clean.head()
```

```python
print("Kepler clean shape:", df_kepler_clean.shape)
print("TESS clean shape:", df_tess_clean.shape)

print("\nDisposition counts (Kepler):")
print(df_kepler_clean["disposition"].value_counts())

print("\nDisposition counts (TESS):")
print(df_tess_clean["disposition"].value_counts())
```

```python
# Drop KP rows before datasets alignment
df_tess_clean = df_tess_clean[df_tess_clean["disposition"] != "KP"]

tess_map = {
    "PC": "CANDIDATE",
    "APC": "CANDIDATE",
    "CP": "CONFIRMED",
    "FP": "FALSE POSITIVE",
    "FA": "FALSE POSITIVE"
}

df_tess_clean["disposition_aligned"] = df_tess_clean["disposition"].map(tess_map)
print(df_tess_clean["disposition_aligned"].value_counts())
```
